```python
def search_by_name(name):
    
    result = df[df["Employee Name"].str.contains(name,
                                                 case=False,
                                                 na=False)]

    if result.empty:
        print("Employee Not Found")
    else:
        return result
```

```python
def search_by_ehrms(code):

    result = df[df["eHrms Code"] == code]

    if result.empty:
        print("Employee Not Found")
    else:
        return result
```

```python
def search_by_office(code):

    result = df[df["Office Code"].astype(str).str.contains(str(code),
                                                           case=False,
                                                           na=False)

               ]

    if result.empty:
        print("Office Not Found")

    else:
        return result
```

```python
def search_employee():

    print("="*60)
    print("EMPLOYEE SEARCH")
    print("="*60)

    print("1. Search by Employee Name")
    print("2. Search by eHrms Code")
    print("3. Search by Office Code")

    choice = input("\nEnter Choice : ")

    if choice == "1":

        name = input("Enter Employee Name : ")

        result = search_by_name(name)

        return result

    elif choice == "2":

        code = int(input("Enter eHrms Code : "))

        result = search_by_ehrms(code)

        return result

    elif choice == "3":

        office = input("Enter Office Code : ")

        result = search_by_office(office)

        return result

    else:

        print("Invalid Choice")
```

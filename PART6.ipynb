```python
import os
from datetime import datetime

import pandas as pd
```

```python

excel_path = r"C:/Users/HP/Downloads/Salary_March 2026 revised (1) By Dam Sir 23.5.26-Final (1).xlsx"

df = pd.read_excel(excel_path)

print("="*60)
print("Employee Database Loaded Successfully")
print("="*60)
```

```python
employee_name = input("Enter Employee Name : ").strip()

employee = df[
    df["Employee Name"].str.contains(
        employee_name,
        case=False,
        na=False
    )
]

if employee.empty:
    raise Exception("Employee Not Found!")

selected_employee = employee.iloc[0]

print("\nEmployee Selected Successfully\n")

print(selected_employee[
    [
        "Employee Name",
        "Designation",
        "Posting District",
        "Place of Posting",
        "Service Tenure(in year)"
    ]
])
```

```python
MINIMUM_EXPERIENCE = 5

experience = int(selected_employee["Service Tenure(in year)"])

if experience >= MINIMUM_EXPERIENCE:

    print("\nEmployee Eligible for Permanent Employment Certificate")

else:

    raise Exception(
        f"Employee has only {experience} years of service.\nMinimum required experience is {MINIMUM_EXPERIENCE} years."
    )
```

```python
os.makedirs("logs", exist_ok=True)

log_file = "logs/certificate_log.xlsx"

print("Logs Folder Ready")
```

```python
if os.path.exists(log_file):

    certificate_log = pd.read_excel(log_file)

else:

    certificate_log = pd.DataFrame(columns=[

        "Certificate Number",

        "Employee Name",

        "eHrms Code",

        "Designation",

        "Posting District",

        "Place of Posting",

        "Experience",

        "Issue Date",

        "Certificate Status",

        "Generated File"

    ])

print("Certificate Log Ready")
```

```python
ehrms = selected_employee["eHrms Code"]

duplicate = certificate_log[
    certificate_log["eHrms Code"] == ehrms
]

if not duplicate.empty:

    raise Exception(
        "Certificate has already been generated for this employee."
    )

print("No Duplicate Certificate Found")
```

```python

today = datetime.today()

certificate_number = f"PEC-{today.year}-{len(certificate_log)+1:04d}"

issue_date = today.strftime("%d-%m-%Y")

print("\nCertificate Number :", certificate_number)
print("Issue Date :", issue_date)
```

```python


filename = (
    selected_employee["Employee Name"]
    .replace(" ", "_")
    + "_Certificate.png"
)

new_record = {

    "Certificate Number": certificate_number,

    "Employee Name": selected_employee["Employee Name"],

    "eHrms Code": selected_employee["eHrms Code"],

    "Designation": selected_employee["Designation"],

    "Posting District": selected_employee["Posting District"],

    "Place of Posting": selected_employee["Place of Posting"],

    "Experience": selected_employee["Service Tenure(in year)"],

    "Issue Date": issue_date,

    "Certificate Status": "Generated",

    "Generated File": filename

}
```

```python


certificate_log = pd.concat(

    [

        certificate_log,

        pd.DataFrame([new_record])

    ],

    ignore_index=True

)

print("Certificate Record Added")
```

```python


certificate_log.to_excel(

    log_file,

    index=False

)

print("="*60)
print("Certificate Log Updated Successfully")
print("="*60)
```

```python
from PIL import Image
from PIL import ImageDraw
from PIL import ImageFont
import os
```

```python

folders = [

    "assets",
    "fonts",
    "output"

]

for folder in folders:

    os.makedirs(folder, exist_ok=True)

print("Project folders created successfully.")
```

```python
WIDTH = 3508
HEIGHT = 2480

certificate = Image.new(

    "RGB",

    (WIDTH, HEIGHT),

    "#FFFDF8"

)

draw = ImageDraw.Draw(certificate)

print("Certificate canvas created.")
```

```python
BACKGROUND = "#FFFDF8"

PRIMARY = "#2F3E46"

GOLD = "#B8860B"

LIGHT_GOLD = "#D6B25E"

DARK = "#1A1A1A"

GRAY = "#666666"

print("Theme loaded.")
```

```python
try:

    TITLE_FONT = ImageFont.truetype(

        "fonts/timesbd.ttf",

        90

    )

    SUBTITLE_FONT = ImageFont.truetype(

        "fonts/times.ttf",

        44

    )

    HEADING_FONT = ImageFont.truetype(

        "fonts/timesbd.ttf",

        60

    )

    TEXT_FONT = ImageFont.truetype(

        "fonts/times.ttf",

        38

    )

    SMALL_FONT = ImageFont.truetype(

        "fonts/times.ttf",

        28

    )

except:

    TITLE_FONT = ImageFont.load_default()

    SUBTITLE_FONT = ImageFont.load_default()

    HEADING_FONT = ImageFont.load_default()

    TEXT_FONT = ImageFont.load_default()

    SMALL_FONT = ImageFont.load_default()

print("Fonts loaded.")
```

```python
# ==========================================================
# Employee Information Section
# ==========================================================

# Section Title
draw.text(
    (500, 1320),
    "EMPLOYEE INFORMATION",
    fill=PRIMARY,
    font=HEADING_FONT
)

# Decorative Line
draw.line(
    [(500,1385),(3000,1385)],
    fill=GOLD,
    width=3
)
```

```python
# ==========================================================
# Employee Information Placeholders
# ==========================================================

labels = [

    "Employee Name",

    "Designation",

    "Office / Health Centre",

    "Posting District",

    "Date of Joining",

    "Years of Service",

    "Employment Status"

]

y = 1470

gap = 95

for label in labels:

    draw.text(

        (520, y),

        f"{label}",

        fill=DARK,

        font=TEXT_FONT

    )

    draw.line(

        [(1300, y+45),

         (2950, y+45)],

        fill=GRAY,

        width=2

    )

    y += gap
```

```python
# ==========================================================
# Eligibility Statement
# ==========================================================

statement = """

This certificate is issued after verification of the employee's
service records maintained by the department.

The employee satisfies the prescribed eligibility requirements
for permanent employment based on completed years of service.

"""

draw.multiline_text(

    (500, 2120),

    statement,

    fill=DARK,

    font=SMALL_FONT,

    spacing=8

)
```

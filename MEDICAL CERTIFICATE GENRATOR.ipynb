```python
import pandas as pd

excel_path = "C:/Users/HP/Downloads/Salary_March 2026 revised (1) By Dam Sir 23.5.26-Final (1).xlsx"

df = pd.read_excel(excel_path)

print("Employee Database Loaded Successfully")
```

```python
print("Total Employees :", len(df))

print("\nEmployee Status\n")
print(df["Status"].value_counts())

print("\nQualification\n")
print(df["Qualification"].value_counts())

print("\nDesignation\n")
print(df["Designation"].value_counts())
```

```python
date_columns = [
    "Appointment Date",
    "Date of Joining"
]

for col in date_columns:
    df[col] = pd.to_datetime(df[col], errors="coerce")

print("Date Conversion Completed")
```

```python
today = datetime.today()

df["Calculated Experience"] = (
    (today - df["Date of Joining"]).dt.days / 365.25
).round(1)

df[["Employee Name", "Calculated Experience"]].head()
```

```python
df.to_excel(
    "database/employee_database_cleaned.xlsx",
    index=False
)

print("Cleaned Database Saved Successfully")
```

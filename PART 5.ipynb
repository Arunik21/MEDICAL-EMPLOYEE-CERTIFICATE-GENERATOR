```python
import os
from datetime import datetime

import pandas as pd

from PIL import Image
from PIL import ImageDraw
from PIL import ImageFont
```

```python
excel_path = r"C:/Users/HP/Downloads/Salary_March 2026 revised (1) By Dam Sir 23.5.26-Final (1).xlsx"

df = pd.read_excel(excel_path)

print("Employee Database Loaded Successfully")
```

```python
employee_name = input("Enter Employee Name : ")

employee = df[
    df["Employee Name"].str.contains(
        employee_name,
        case=False,
        na=False
    )
]

if employee.empty:

    raise Exception("Employee Not Found!")

selected_employee = employee.iloc[0]

print("Employee Selected Successfully")
```

```python
MINIMUM_EXPERIENCE = 5

experience = selected_employee["Service Tenure(in year)"]

if experience < MINIMUM_EXPERIENCE:

    raise Exception("Employee is not eligible for Permanent Employment Certificate.")

print("Employee Eligible")
```

```python
certificate = Image.open(
    "output/master_certificate_template.png"
)

draw = ImageDraw.Draw(certificate)

print("Certificate Template Loaded")
```

```python
TITLE_FONT = ImageFont.truetype(
    "fonts/timesbd.ttf",
    52
)

TEXT_FONT = ImageFont.truetype(
    "fonts/times.ttf",
    40
)

SMALL_FONT = ImageFont.truetype(
    "fonts/times.ttf",
    30
)
```

```python
current_year = datetime.now().year

ehrms = selected_employee["eHrms Code"]

certificate_number = f"PEC-{current_year}-{ehrms}"

issue_date = datetime.today().strftime("%d-%m-%Y")

print(certificate_number)
```

```python
draw.text(

    (720,770),

    certificate_number,

    font=TEXT_FONT,

    fill="black"

)
```

```python
draw.text(

    (2870,770),

    issue_date,

    font=TEXT_FONT,

    fill="black"

)
```

```python
employee_information = [

    selected_employee["Employee Name"],

    selected_employee["Designation"],

    selected_employee["Place of Posting"],

    selected_employee["Posting District"],

    pd.to_datetime(
        selected_employee["Date of Joining"],
        errors="coerce"
    ).strftime("%d-%m-%Y"),

    f"{selected_employee['Service Tenure(in year)']} Years",

    selected_employee["Status"]

]
```

```python
y = 1470

gap = 95

for value in employee_information:

    draw.text(

        (1320,y),

        str(value),

        font=TEXT_FONT,

        fill="black"

    )

    y += gap
```

```python
draw.text(

    (1180,1145),

    selected_employee["Employee Name"],

    font=TITLE_FONT,

    fill="black"

)
```

```python
certificate_text = f"""

This is to certify that

{selected_employee['Employee Name']}

has successfully completed

{selected_employee['Service Tenure(in year)']} years

of continuous government service.

The employee fulfills all departmental
eligibility criteria and is hereby recognized
for Permanent Employment.

"""

draw.multiline_text(

    (650,960),

    certificate_text,

    font=TEXT_FONT,

    fill="black",

    spacing=12

)
```

```python
os.makedirs(

    "generated_certificates",

    exist_ok=True

)
```

```python
filename = (

    selected_employee["Employee Name"]

    .replace(" ","_")

    + "_Certificate.png"

)

output_path = os.path.join(

    "generated_certificates",

    filename

)

certificate.save(

    output_path,

    dpi=(300,300)

)

print("="*60)

print("CERTIFICATE GENERATED SUCCESSFULLY")

print(f"Saved at : {output_path}")

print("="*60)
```

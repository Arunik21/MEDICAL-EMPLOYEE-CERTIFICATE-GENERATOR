```python
import os
import pandas as pd

import ipywidgets as widgets

from IPython.display import display
from IPython.display import clear_output

from datetime import datetime
```

```python


excel_path = r"C:/Users/HP/Downloads/Salary_March 2026 revised (1) By Dam Sir 23.5.26-Final (1).xlsx"

df = pd.read_excel(excel_path)

print("Employee Database Loaded Successfully")
```

```python
title = widgets.HTML("""

<h2 style="text-align:center;color:#003366;">

Medical Certificate Generator

</h2>

""")

display(title)
```

```python
employee_box = widgets.Text(

    placeholder="Enter Employee Name",

    description="Employee:",

    layout=widgets.Layout(width="550px")

)

display(employee_box)
```

```python
search_button = widgets.Button(

    description="Search Employee",

    button_style="info",

    icon="search"

)

eligibility_button = widgets.Button(

    description="Check Eligibility",

    button_style="success",

    icon="check"

)

generate_button = widgets.Button(

    description="Generate Certificate",

    button_style="warning",

    icon="file"

)

display(search_button)

display(eligibility_button)

display(generate_button)
```

```python
def search_employee(button):

    global selected_employee

    with output:

        clear_output()

        employee_name = employee_box.value.strip()

        employee = df[

            df["Employee Name"]

            .str.contains(

                employee_name,

                case=False,

                na=False

            )

        ]

        if employee.empty:

            print("Employee Not Found")

            return

        selected_employee = employee.iloc[0]

        print("="*50)

        print("Employee Found Successfully")

        print("="*50)

        print()

        print("Name :", selected_employee["Employee Name"])

        print("Designation :", selected_employee["Designation"])

        print("Posting District :", selected_employee["Posting District"])

        print("Place of Posting :", selected_employee["Place of Posting"])

        print("Experience :", selected_employee["Service Tenure(in year)"], "Years")
```

```python
search_button.on_click(search_employee)
```

```python
def check_eligibility(button):

    with output:

        clear_output()

        try:

            experience = int(

                selected_employee["Service Tenure(in year)"]

            )

        except NameError:

            print("Search an employee first.")

            return

        print()

        print("Experience :", experience, "Years")

        print()

        if experience >= 5:

            print("Employee is Eligible")

        else:

            print("Employee is NOT Eligible")
```

```python
eligibility_button.on_click(check_eligibility)
```

```python
def generate_certificate(button):

    with output:

        clear_output()

        try:

            experience = int(

                selected_employee["Service Tenure(in year)"]

            )

        except NameError:

            print("Search an employee first.")

            return

        if experience < 10:

            print("Employee Not Eligible")

            return

        print("Generating Certificate...")

        print()

        print("Employee :", selected_employee["Employee Name"])

        print("Designation :", selected_employee["Designation"])

        print("Certificate Number :")

        print(

            f"PEC-{datetime.now().year}-{selected_employee['eHrms Code']}"

        )

        print()

        print("Certificate Generated Successfully")

        print()

        print("Saved inside generated_certificates/")
```

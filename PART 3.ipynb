### Experience Calculation & Eligibility Module

```python
import pandas as pd
from datetime import datetime
```

```python

excel_path = r"C:/Users/HP/Downloads/Salary_March 2026 revised (1) By Dam Sir 23.5.26-Final (1).xlsx"

df = pd.read_excel(excel_path)

print("Employee Database Loaded Successfully")
print("Total Employees :", len(df))
```

```python
date_columns = [
    "Appointment Date",
    "Date of Joining",
    "Date of Retirement"
]

for column in date_columns:
    df[column] = pd.to_datetime(
        df[column],
        errors="coerce"
    )

print("Date Columns Converted Successfully")
```

```python

employee_name = input("Enter Employee Name : ")

employee = df[
    df["Employee Name"].str.contains(
        employee_name,
        case=False,
        na=False
    )
]

if employee.empty:

    print("Employee Not Found")

else:

    print("Employee Found")
```

```python
selected_employee = employee.iloc[0]

print(selected_employee["Employee Name"])
```

```python
MINIMUM_EXPERIENCE = 10

ACTIVE_STATUS = [
    "working",
    "active"
]
```

```python

def calculate_experience(joining_date):

    if pd.isnull(joining_date):
        return None

    today = datetime.today()

    experience = (today - joining_date).days / 365.25

    return round(experience,1)
```

```python
def get_employee_experience(employee):

    tenure = employee["Service Tenure(in year)"]

    if pd.notnull(tenure):

        return float(tenure)

    return calculate_experience(
        employee["Date of Joining"]
    )
```

```python
def is_employee_active(employee):

    status = str(
        employee["Status"]
    ).strip().lower()

    return status in ACTIVE_STATUS
```

```python
def check_eligibility(employee):

    experience = get_employee_experience(employee)

    if experience >= MINIMUM_EXPERIENCE and is_employee_active(employee):

        return "Eligible"

    elif experience >= MINIMUM_EXPERIENCE:

        return "Inactive Employee"

    else:

        return "Not Eligible"
```

```python
def employee_summary(employee):

    print("="*60)

    print("EMPLOYEE SUMMARY")

    print("="*60)

    print("Employee Name      :", employee["Employee Name"])
    print("Designation        :", employee["Designation"])
    print("Posting District   :", employee["Posting District"])
    print("Place of Posting   :", employee["Place of Posting"])
    print("Date of Joining    :", employee["Date of Joining"].date())
    print("Experience         :", get_employee_experience(employee),"Years")
    print("Status             :", employee["Status"])
    print("Eligibility        :", check_eligibility(employee))

    print("="*60)
```

```python
employee_summary(selected_employee)
```

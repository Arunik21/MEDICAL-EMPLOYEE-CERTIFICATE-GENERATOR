```python
import os
from datetime import datetime

import pandas as pd
import qrcode

from PIL import Image
from PIL import ImageDraw
from PIL import ImageFont

print("=" * 60)
print("PART 8 INITIALIZED SUCCESSFULLY")
print("All Required Libraries Imported.")
print("=" * 60)
```

```python
# ==========================================================
# PROJECT CONFIGURATION
# ==========================================================

# Employee Database

EXCEL_PATH = r"C:\Users\HP\Downloads\Salary_March 2026 revised (1) By Dam Sir 23.5.26-Final (1).xlsx"

# Certificate Template

TEMPLATE_PATH = r"D:\INTERNSHIP PROJECT\output\certificate_template_final.png"

# Output Directories

CERTIFICATE_FOLDER = r"D:\INTERNSHIP PROJECT\Generated Certificates"

QR_FOLDER = r"D:\INTERNSHIP PROJECT\QR Codes"

LOG_FOLDER = r"D:\INTERNSHIP PROJECT\Certificate Logs"

FONT_FOLDER = r"C:\Windows\Fonts"

# ==========================================================
# Create Output Directories
# ==========================================================

os.makedirs(CERTIFICATE_FOLDER, exist_ok=True)
os.makedirs(QR_FOLDER, exist_ok=True)
os.makedirs(LOG_FOLDER, exist_ok=True)

print("Project Configuration Loaded Successfully.")
```

```python
# ==========================================================
# LOAD EMPLOYEE DATABASE
# ==========================================================

try:

    df = pd.read_excel(EXCEL_PATH)

    # Remove extra spaces from column names

    df.columns = df.columns.str.strip()

    print()

    print("Employee Database Loaded Successfully")

    print("Total Employees :", len(df))

except Exception as error:

    raise Exception(

        f"Unable to Load Employee Database\n\n{error}"

    )
```

```python
# ==========================================================
# SEARCH EMPLOYEE
# ==========================================================

employee_name = input("Enter Employee Name : ").strip()

employee = df[

    df["Employee Name"]

    .astype(str)

    .str.strip()

    .str.lower()

    == employee_name.lower()

]

if employee.empty:

    raise Exception("Employee Not Found.")

selected_employee = employee.iloc[0]

print()

print("=" * 60)
print("EMPLOYEE FOUND SUCCESSFULLY")
print("=" * 60)

print(selected_employee)
```

```python
# ==========================================================
# ELIGIBILITY CHECK
# ==========================================================

MINIMUM_EXPERIENCE = 10

experience = float(

    selected_employee["Service Tenure(in year)"]

)

status = str(

    selected_employee["Status"]

).strip().lower()

if experience < MINIMUM_EXPERIENCE:

    raise Exception(

        "Employee is not eligible for Permanent Employment Certificate."

    )



print()

print("=" * 60)

print("EMPLOYEE ELIGIBLE")

print("=" * 60)

print(f"Experience : {experience} Years")

print(f"Status : {status}")
```

```python
# ==========================================================
# GENERATE CERTIFICATE NUMBER
# ==========================================================

certificate_number = (

    "PEC-"

    + datetime.now().strftime("%Y%m%d%H%M%S")

)

generation_date = datetime.now().strftime("%d-%m-%Y")

generation_time = datetime.now().strftime("%H:%M:%S")

print("=" * 60)

print("CERTIFICATE NUMBER GENERATED")

print("=" * 60)

print("Certificate Number :", certificate_number)

print("Generation Date :", generation_date)

print("Generation Time :", generation_time)
```

```python
# ==========================================================
# GENERATE QR CODE
# ==========================================================

qr_information = f"""
Permanent Employment Certificate

Certificate Number : {certificate_number}

Employee Name : {selected_employee['Employee Name']}

eHrms Code : {selected_employee['eHrms Code']}

Designation : {selected_employee['Designation']}

Place of Posting : {selected_employee['Place of Posting']}

Posting District : {selected_employee['Posting District']}

Date of Joining : {selected_employee['Date of Joining']}

Service Tenure : {selected_employee['Service Tenure(in year)']} Years

Status : {selected_employee['Status']}

Generated On : {generation_date}

Generated At : {generation_time}
"""

qr = qrcode.QRCode(

    version=1,

    error_correction=qrcode.constants.ERROR_CORRECT_H,

    box_size=8,

    border=4

)

qr.add_data(qr_information)

qr.make(fit=True)

qr_image = qr.make_image(fill_color="black", back_color="white")

qr_filename = os.path.join(

    QR_FOLDER,

    f"{certificate_number}.png"

)

qr_image.save(qr_filename)

print("=" * 60)

print("QR CODE GENERATED SUCCESSFULLY")

print("=" * 60)

print(qr_filename)
```

```python
# ==========================================================
# LOAD CERTIFICATE TEMPLATE
# ==========================================================

if not os.path.exists(TEMPLATE_PATH):

    raise FileNotFoundError(

        f"""

Certificate Template Not Found

{TEMPLATE_PATH}

Please complete Part 4 first.

"""

    )

certificate = Image.open(TEMPLATE_PATH)

draw = ImageDraw.Draw(certificate)

print("=" * 60)

print("CERTIFICATE TEMPLATE LOADED")

print("=" * 60)

print("Template Size :", certificate.size)
```

```python
# ==========================================================
# LOAD FONTS
# ==========================================================

font_path = os.path.join(

    FONT_FOLDER,

    "arial.ttf"

)

bold_font_path = os.path.join(

    FONT_FOLDER,

    "arialbd.ttf"

)

try:

    title_font = ImageFont.truetype(

        bold_font_path,

        44

    )

    heading_font = ImageFont.truetype(

        bold_font_path,

        32

    )

    text_font = ImageFont.truetype(

        font_path,

        28

    )

    small_font = ImageFont.truetype(

        font_path,

        22

    )

except:

    title_font = ImageFont.load_default()

    heading_font = ImageFont.load_default()

    text_font = ImageFont.load_default()

    small_font = ImageFont.load_default()

print("=" * 60)

print("FONTS LOADED SUCCESSFULLY")

print("=" * 60)
```

```python
# ==========================================================
# PREPARE EMPLOYEE INFORMATION
# ==========================================================

joining_date = pd.to_datetime(

    selected_employee["Date of Joining"],

    errors="coerce"

)

if pd.isna(joining_date):

    joining_date = "N/A"

else:

    joining_date = joining_date.strftime("%d-%m-%Y")

employee_information = {

    "Employee Name":

        selected_employee["Employee Name"],

    "Designation":

        selected_employee["Designation"],

    "Place of Posting":

        selected_employee["Place of Posting"],

    "Posting District":

        selected_employee["Posting District"],

    "Date of Joining":

        joining_date,

    "Service Tenure":

        f"{selected_employee['Service Tenure(in year)']} Years",

    "Status":

        selected_employee["Status"],

    "Certificate Number":

        certificate_number

}

print("=" * 60)

print("EMPLOYEE INFORMATION READY")

print("=" * 60)

for key, value in employee_information.items():

    print(f"{key} : {value}")
```

```python
# ==========================================================
# WRITE EMPLOYEE DETAILS ON CERTIFICATE
# ==========================================================

# Employee Name
draw.text(
    (520, 470),
    str(selected_employee["Employee Name"]),
    fill="black",
    font=text_font
)

# Designation
draw.text(
    (520, 530),
    str(selected_employee["Designation"]),
    fill="black",
    font=text_font
)

# Place of Posting
draw.text(
    (520, 590),
    str(selected_employee["Place of Posting"]),
    fill="black",
    font=text_font
)

# Posting District
draw.text(
    (520, 650),
    str(selected_employee["Posting District"]),
    fill="black",
    font=text_font
)

# Date of Joining
draw.text(
    (520, 710),
    joining_date,
    fill="black",
    font=text_font
)

# Service Tenure
draw.text(
    (520, 770),
    f"{selected_employee['Service Tenure(in year)']} Years",
    fill="black",
    font=text_font
)

# Status
draw.text(
    (520, 830),
    str(selected_employee["Status"]),
    fill="black",
    font=text_font
)

# Certificate Number
draw.text(
    (520, 890),
    certificate_number,
    fill="black",
    font=text_font
)

print("Employee Information Written Successfully.")
```

```python
# ==========================================================
# INSERT QR CODE
# ==========================================================

qr_image = Image.open(qr_filename)

qr_image = qr_image.resize((180, 180))

certificate.paste(
    qr_image,
    (1250, 730)
)

# Generated Date
draw.text(
    (1050, 950),
    generation_date,
    fill="black",
    font=small_font
)

print("QR Code Added Successfully.")
```
